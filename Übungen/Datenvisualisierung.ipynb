```python
import matplotlib.pyplot as plt
import numpy as np
# Daten für die folgenden Aufgaben
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)


ax1 = plt.plot(x, y, label='Sin(x)', color="blue")
ax2 = plt.plot(x, z, label='Cos(x)',linestyle='dashed', color="red")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("IDFK")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Daten für die folgenden Aufgaben
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)

fig = plt.figure(figsize=(10, 15))

ax1 = plt.subplot2grid((3, 2), (0, 0))
ax1 = plt.plot(x, y, label='Sin(x)', color="blue")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("IDFK")


ax2 = plt.subplot2grid((3, 2), (0, 1))
ax2 = plt.plot(x, z, label='Cos(x)',linestyle='dashed', color="red")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("IDFK")


plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Daten für die folgenden Aufgaben
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)


ax1 = plt.plot(x, y, label='Sin(x)', color="blue")
ax2 = plt.plot(x, z, label='Cos(x)',linestyle='dashed', color="red")


plt.fill_between(x,z,y)


plt.xlabel("X")
plt.ylabel("Y")
plt.title("IDFK")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(19680801)
N_points = 1000
n_bins = 20

# Generate two normal distributions
dist1 = rng.standard_normal(N_points)


axs = plt.hist(dist1, bins=n_bins)

# We can set the number of bins with the *bins* keyword argument.

plt.show()
```
