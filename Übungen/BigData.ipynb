```python
data1 = ["Green", "Blue", "Blue", "Orange"]
data2 = ["Green", "Red", "Blue", "Orange"]


idfk ={}

def add_to_map(arr, dic):
    for x in arr:
        if x in dic.keys():
            dic[x] += 1
        else:
            dic[x] = 1

add_to_map(data1, idfk)
add_to_map(data2, idfk)


idfk

```
